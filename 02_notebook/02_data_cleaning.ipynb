```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("../01_Dataset/01_raw/Heart_raw_uncleaned.csv")
df.head()
```

**COMFIRMING THE MISSING VALUES**

```python
df.isnull().sum()
```

```python
df["BMI"] = df["BMI"].fillna(df["BMI"].median()) 
```

```python
df["Smoking"] = df["Smoking"].fillna(df["Smoking"].mode()[0])
```

```python
df["PhysicalHealth"] = df["PhysicalHealth"].fillna(df["PhysicalHealth"].mode()[0])
```

```python
df["GenHealth"] = df["GenHealth"].fillna(df["GenHealth"].mode()[0])
```

```python
df["SleepTime"] = df["SleepTime"].fillna(df["SleepTime"].median())
```

```python
df.isnull().sum()
```

**REMOVING THE DUPLICATE VALUES**

```python
df = df.drop_duplicates()
```
