**PATIENT DEMOGRAPHIC AND HEART ANALYSIS**

**CALCUATING THE PERCENTAGE OF AGE CATE. VS HEART DIS.**

```python
age_total = df["AgeCategory"].value_counts().sort_index()
age_heart_disease = df[df["HeartDisease"] == "Yes"]["AgeCategory"].value_counts().sort_index()
age_percentage = (age_heart_disease / age_total) * 100
age_percentage = age_percentage.sort_index()
```
